## 0. Celda de arranque (ejecútala primero, cada vez)

En Colab, la sesión se borra al cerrarla o al desconectarse por inactividad. Esta celda trae de nuevo los archivos de la plantilla (la carpeta `src/` y los datos de práctica) y deja las rutas listas. Tarda unos segundos. En una instalación local no hace nada.

```python
# CELDA DE ARRANQUE: ejecútala primero en cada sesión de Colab.
import os
import subprocess
import sys

REPO_URL = "https://github.com/USUARIO/seminario-python-encuestas.git"  # el docente indica el enlace real
CARPETA = "/content/seminario"

if "google.colab" in sys.modules:
    if "USUARIO" in REPO_URL:
        raise RuntimeError("Falta el enlace de la plantilla: el docente lo indicará.")
    if not os.path.exists(CARPETA):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, CARPETA], check=True)
    os.chdir(os.path.join(CARPETA, "notebooks"))
    print("Plantilla lista en Colab:", os.getcwd())
else:
    print("Instalación local: no hace falta arrancar nada.")
```

## 4. Una celda de prueba

Calcula una media y dibuja un gráfico para comprobar que todo funciona junto.

```python
import matplotlib.pyplot as plt

x = [10, 20, 30, 40]
print("Media:", sum(x) / len(x))

plt.plot(x, marker="o")
plt.title("Mi primer gráfico")
plt.xlabel("Posición (empieza en 0 en Python)")
plt.ylabel("Valor")
plt.show()
```
